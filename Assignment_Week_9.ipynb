```python
import requests
import pandas as pd

url ="https://dummyjson.com/products"

products = requests.get(url)

products.status_code

products_data = products.json()

print(type(products_data))

products_df = pd.DataFrame(products_data)
print (products_df.head())

products_csv = products_df.to_csv("products_df.csv", index = False)
```
